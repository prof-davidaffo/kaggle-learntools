## 2. Variabili e assegnazione

Una **variabile** è un nome associato a un valore. L’istruzione di **assegnazione**, indicata dal simbolo `=`, stabilisce questa associazione.

Nell’istruzione `price = 200000`, il nome `price` viene associato al numero intero `200000`. Il valore può essere richiamato nelle istruzioni successive attraverso il nome della variabile.

Gli identificatori degli esempi seguono le convenzioni dei notebook del corso: `price` indica il prezzo e `area` la superficie. Il prezzo al metro quadrato si ottiene dividendo il primo valore per il secondo.


````python
price = 200000
area = 80
price_per_square_meter = price / area

print(price)
print(price_per_square_meter)
````

La prima istruzione `print()` visualizza `200000`; la seconda visualizza `2500.0`.

L’assegnazione `price_per_square_meter = price / area` avviene in due passaggi: Python calcola l’espressione a destra di `=`, utilizzando i valori di `price` e `area`, quindi associa il risultato al nome `price_per_square_meter`.

I nomi delle variabili possono contenere lettere, cifre e il carattere `_`; il primo carattere deve essere una lettera oppure `_`. Per separare le parole si usa comunemente il trattino basso, come in `price_per_square_meter`. Maiuscole e minuscole sono distinte: `price` e `Price` identificano due nomi diversi.

Una nuova assegnazione aggiorna il valore associato alla variabile.


````python
rooms = 3
print(rooms)
rooms = 4
print(rooms)
````

Le due stampe producono rispettivamente `3` e `4`.

Moltiplicazione e divisione hanno precedenza su addizione e sottrazione. Le parentesi tonde permettono di stabilire un diverso ordine di calcolo: `(10 + 3) * 2` vale `26`, mentre `10 + 3 * 2` vale `16`.

I numeri decimali si scrivono con il punto, per esempio `82.5`. Negli esempi i numeri interi sono scritti con le sole cifre, come `200000`.

## 3. Tipi di dato e confronti

Il **tipo** di un dato determina le operazioni che si possono eseguire su di esso. I tipi di base utilizzati negli esempi sono i seguenti.

| Tipo | Descrizione | Esempio |
| --- | --- | --- |
| `int` | Numero intero | `3` |
| `float` | Numero in virgola mobile, usato per rappresentare valori decimali | `82.5` |
| `str` | Sequenza di caratteri, detta **stringa** | `"Prezzo"` |
| `bool` | Valore **booleano**, vero oppure falso | `True`, `False` |

Le virgolette identificano una stringa: `80` è un numero intero, mentre `"80"` è una stringa di due caratteri. Il significato di un operatore dipende dal tipo dei valori a cui viene applicato.


````python
print(80 + 20)
print("80" + "20")
````

La prima espressione produce `100`, risultato dell’addizione tra interi. La seconda produce `8020`: sulle stringhe l’operatore `+` esegue la **concatenazione**, cioè unisce i due testi nell’ordine indicato.

La funzione `type()` restituisce il tipo di un valore. Per esempio, `print(type(80))` visualizza `<class 'int'>`.

### Operatori di confronto

Un confronto restituisce un valore booleano: `True` quando la relazione è verificata, `False` nel caso contrario.

| Relazione | Operatore | Espressione | Risultato |
| --- | --- | --- | --- |
| Uguaglianza | `==` | `3 == 3` | `True` |
| Diversità | `!=` | `3 != 3` | `False` |
| Minore | `<` | `3 < 5` | `True` |
| Minore o uguale | `<=` | `3 <= 3` | `True` |
| Maggiore | `>` | `3 > 5` | `False` |
| Maggiore o uguale | `>=` | `3 >= 3` | `True` |

L’istruzione `rooms = 3` assegna un valore; l’espressione `rooms == 3` confronta il valore della variabile con il numero 3.


````python
price = 180000
budget = 200000
within_budget = price <= budget

print(within_budget)
````

Il confronto `price <= budget` restituisce `True`, poiché 180000 è minore o uguale a 200000. Il risultato viene assegnato alla variabile `within_budget`.

## 4. Istruzione condizionale e indentazione

L’istruzione **condizionale** permette di scegliere quali operazioni eseguire in base al valore di una condizione. La forma `if ... else` prevede due casi: il blocco associato a `if` viene eseguito quando la condizione è vera; quello associato a `else` viene eseguito quando è falsa.

Il programma seguente confronta il prezzo di una casa con il budget disponibile.


````python
price = 180000
budget = 200000

if price <= budget:
    print("La casa rientra nel budget")
else:
    print("La casa supera il budget")

print("Controllo terminato")
````

Con i valori assegnati, il programma produce:

```text
La casa rientra nel budget
Controllo terminato
```

La struttura dell’istruzione si legge nel modo seguente:

1. `if price <= budget:` valuta la condizione. I due punti introducono il blocco di istruzioni associato.
2. Il primo `print()` appartiene al blocco `if` ed è preceduto da quattro spazi.
3. `else:` è allineato con `if` e introduce il blocco del caso alternativo.
4. L’ultimo `print()` si trova al margine sinistro e viene eseguito dopo l’istruzione condizionale, in entrambi i casi.

Gli spazi all’inizio di una riga costituiscono l’**indentazione**, o rientro. Un **blocco** è un gruppo di istruzioni che appartengono alla stessa struttura di controllo. In Python l’indentazione ne delimita il contenuto. In questo corso si usano quattro spazi per ciascun livello di rientro, con allineamento uniforme delle istruzioni dello stesso blocco.

**Attività.** Impostare `price = 250000`. La prima stampa deve diventare `La casa supera il budget`; la stampa finale rimane `Controllo terminato`.

La forma semplice `if`, priva del ramo `else`, esegue il blocco soltanto quando la condizione è vera; successivamente il programma prosegue con l’istruzione che segue il blocco.

## 5. Liste e accesso agli elementi

Una **lista** è una sequenza ordinata di elementi. Si scrive racchiudendo gli elementi tra parentesi quadre e separandoli con virgole.

La lista seguente contiene i prezzi di tre abitazioni.


````python
prices = [150000, 200000, 270000]
print(prices)
print(len(prices))
````

La stampa di `prices` visualizza l’intera lista. La funzione `len(prices)` restituisce il numero di elementi, pari a `3`.

### Indici

L’**indice** identifica la posizione di un elemento nella lista. La numerazione parte da zero.

| Posizione | Prima | Seconda | Terza |
| --- | --- | --- | --- |
| Indice | `0` | `1` | `2` |
| Valore | `150000` | `200000` | `270000` |

L’espressione `prices[indice]` restituisce l’elemento nella posizione specificata.


````python
print(prices[0])  # Primo prezzo: 150000.
print(prices[1])  # Secondo prezzo: 200000.
print(prices[2])  # Terzo prezzo: 270000.
````

Per una lista di tre elementi gli indici positivi validi sono `0`, `1` e `2`. L’accesso `prices[3]` provoca un errore `IndexError`, poiché la posizione richiesta supera quelle disponibili.

Gli indici negativi contano a partire dalla fine: `prices[-1]` restituisce l’ultimo elemento.

### Selezione di un intervallo

La scrittura `[inizio:fine]` seleziona una parte della lista. L’indice iniziale è incluso nell’intervallo; l’indice finale è escluso. L’operazione prende il nome di **slicing**.

Nell’esempio seguente, `prices[0:2]` seleziona gli elementi di indice `0` e `1`.


````python
first_two_prices = prices[0:2]
print(first_two_prices)
````

Il risultato è `[150000, 200000]`.

L’omissione dell’indice iniziale estende la selezione all’inizio della lista; l’omissione dell’indice finale la estende fino alla fine.

| Espressione | Intervallo selezionato | Risultato |
| --- | --- | --- |
| `prices[:2]` | Dall’inizio fino all’indice 2 escluso | `[150000, 200000]` |
| `prices[1:]` | Dall’indice 1 incluso fino alla fine | `[200000, 270000]` |
| `prices[:]` | Tutti gli elementi, in una nuova lista | `[150000, 200000, 270000]` |

## 6. Ciclo `for`

Un **ciclo** permette di eseguire ripetutamente un blocco di istruzioni. Il ciclo `for` scorre gli elementi di una sequenza, nell’ordine in cui compaiono.

Nell’istruzione `for price in prices:`, la variabile `price` assume a ogni ripetizione il valore dell’elemento successivo di `prices`. Per ciascun valore viene eseguito il blocco rientrato.


````python
prices = [150000, 250000, 180000]

for price in prices:
    print("Prezzo:", price)

print("Elenco terminato")
````

La funzione `print()` può ricevere più valori separati da virgole. Nell’espressione `print("Prezzo:", price)` visualizza il testo e il valore della variabile, separati da uno spazio.

L’esecuzione del ciclo si articola in tre ripetizioni:

| Ripetizione | Valore di `price` | Stampa |
| --- | --- | --- |
| Prima | `150000` | `Prezzo: 150000` |
| Seconda | `250000` | `Prezzo: 250000` |
| Terza | `180000` | `Prezzo: 180000` |

L’istruzione `print("Elenco terminato")` si trova fuori dal blocco e viene eseguita una sola volta, al termine del ciclo.

### Blocchi annidati

Un blocco può contenere un’altra struttura di controllo. Si parla in questo caso di **annidamento**. Il programma seguente verifica, per ogni prezzo, la condizione `price <= 200000`.


````python
for price in prices:
    if price <= 200000:
        print("Nel budget:", price)

print("Controlli terminati")
````

Risultato:

```text
Nel budget: 150000
Nel budget: 180000
Controlli terminati
```

L’istruzione `if` appartiene al blocco del `for` e ha un rientro di quattro spazi. Il `print()` condizionale appartiene al blocco dell’`if` e ha un rientro di otto spazi. La stampa finale è esterna a entrambi i blocchi.

## 7. Funzioni

Una **funzione** raccoglie un insieme di operazioni richiamabili attraverso un nome. La **chiamata** di una funzione si scrive con il nome seguito da parentesi tonde; i valori passati nelle parentesi sono gli **argomenti**.

Tra le funzioni predefinite di Python figurano:

- `print()`, che visualizza i valori ricevuti;
- `sum()`, che calcola la somma degli elementi di una sequenza numerica;
- `len()`, che restituisce il numero di elementi di una sequenza.

Il valore **restituito** da una funzione può essere assegnato a una variabile oppure utilizzato in un’espressione. Nell’esempio seguente somma e numero di elementi vengono utilizzati per calcolare la media aritmetica.


````python
values = [10, 20, 30]
total = sum(values)
count = len(values)
print(total)
print(count)
print(total / count)
````

Le stampe producono `60`, `3` e `20.0`. La media è il rapporto tra la somma dei valori e il loro numero.

### Definizione di una funzione

L’istruzione `def` permette di definire una funzione. La funzione `unit_cost` dell’esempio seguente calcola il prezzo al metro quadrato a partire da prezzo e superficie.


````python
def unit_cost(price, area):
    return price / area
````

La definizione comprende il nome della funzione, l’elenco dei **parametri** e il corpo rientrato. I parametri `price` e `area` sono i nomi con cui la funzione si riferisce ai valori ricevuti durante una chiamata.

L’istruzione `return` termina l’esecuzione della funzione e restituisce il valore dell’espressione che la segue. In questo caso il risultato è `price / area`.

L’esecuzione della cella registra la definizione di `unit_cost`. Il calcolo viene eseguito in occasione di ciascuna chiamata, con gli argomenti specificati.


````python
first_cost = unit_cost(200000, 80)
second_cost = unit_cost(270000, 120)
print(first_cost)
print(second_cost)
````

Le due chiamate restituiscono rispettivamente `2500.0` e `2250.0`. Nella prima, il parametro `price` riceve `200000` e il parametro `area` riceve `80`; nella seconda ricevono `270000` e `120`.

Gli argomenti possono essere passati per posizione oppure per nome. Le due chiamate seguenti sono equivalenti.


````python
print(unit_cost(200000, 80))
print(unit_cost(price=200000, area=80))
````

Nella seconda chiamata, `price=200000` e `area=80` associano esplicitamente ogni argomento al parametro corrispondente. Questa scrittura è frequente nelle funzioni delle librerie utilizzate nel corso.

### Visualizzazione e restituzione di un valore

`print()` visualizza un valore sullo schermo. `return` rende disponibile il risultato della funzione al codice che l’ha chiamata. Nell’esempio, il risultato di `unit_cost()` viene conservato in `first_cost` e può essere utilizzato in altri calcoli.

Una funzione che termina senza un’istruzione `return` restituisce `None`. Anche `print()` restituisce `None`: la visualizzazione sullo schermo è l’effetto della sua esecuzione.

## 8. Dizionari

Un **dizionario** è una raccolta di associazioni tra **chiavi** e **valori**. L’accesso a un valore avviene mediante la chiave corrispondente.

Nell’esempio, le chiavi `"Rooms"`, `"Area"` e `"Price"` identificano le caratteristiche di una casa. Il dizionario è delimitato da parentesi graffe; i due punti separano ciascuna chiave dal valore associato.


````python
house = {"Rooms": 3, "Area": 80.0, "Price": 200000}
print(house["Area"])

house["Rooms"] = 4
print(house["Rooms"])
````

Le stampe producono `80.0` e `4`. L’espressione `house["Area"]` restituisce il valore associato alla chiave `"Area"`. L’assegnazione a `house["Rooms"]` aggiorna il numero di stanze.

Le chiavi usate nell’esempio sono stringhe e vengono scritte tra virgolette. La distinzione tra maiuscole e minuscole vale anche in questo caso: `"Area"` e `"area"` sono chiavi diverse.

### Riepilogo delle parentesi

| Parentesi | Esempio | Uso |
| --- | --- | --- |
| Tonde `()` | `print(price)` | Chiamata di una funzione |
| Tonde `()` | `(price + 10000) / area` | Raggruppamento di un’espressione |
| Quadre `[]` | `[150000, 200000]` | Creazione di una lista |
| Quadre `[]` | `prices[0]`, `house["Area"]` | Accesso a un elemento |
| Graffe `{}` | `{"Rooms": 3}` | Creazione di un dizionario |

L’istruzione `import` carica un modulo e rende disponibili i suoi strumenti. La clausola `as` permette di scegliere un nome abbreviato. L’istruzione seguente importa pandas con l’abbreviazione convenzionale `pd`.


````python
import pandas as pd
````

Il nome `pd` permette ora di accedere agli strumenti di pandas. La scrittura `pd.DataFrame(...)`, utilizzata nella sezione successiva, richiama lo strumento `DataFrame` attraverso il punto.

In Python i valori elaborati dal programma sono **oggetti**. Un oggetto possiede **attributi**, che ne descrivono le proprietà, e può fornire **metodi**, cioè funzioni associate all’oggetto stesso.

## 10. Tabelle con pandas

Un **dataset** è una raccolta di dati. Nell’esempio seguente ogni riga descrive un’abitazione e ogni colonna una sua caratteristica.

| Abitazione | Rooms: stanze | Area: superficie in m² | Price: prezzo in euro |
| --- | --- | --- | --- |
| Prima | 2 | 60 | 150000 |
| Seconda | 3 | 80 | 200000 |
| Terza | 4 | 120 | 270000 |
| Quarta | 3 | dato mancante | 190000 |

In pandas, un **DataFrame** è una struttura tabellare organizzata in righe e colonne. Per costruirlo si può passare a `pd.DataFrame()` un dizionario: ogni chiave fornisce il nome di una colonna e ogni lista contiene i valori di quella colonna.

Le liste hanno la stessa lunghezza e rispettano l’ordine delle abitazioni. Il valore `None` rappresenta la superficie mancante della quarta casa.


````python
data = pd.DataFrame({
    "Rooms": [2, 3, 4, 3],
    "Area": [60.0, 80.0, 120.0, None],
    "Price": [150000, 200000, 270000, 190000],
})
print(data)
````

Le righe sono numerate da `0` a `3`. Per la superficie della quarta casa pandas visualizza `NaN`, che in questa colonna numerica rappresenta un dato mancante.

Le parentesi aperte consentono di distribuire un’espressione su più righe. I rientri all’interno di `pd.DataFrame({...})` rendono leggibile la struttura del dizionario; il loro ruolo è distinto dall’indentazione dei blocchi di istruzioni.

### 10.1 Esplorazione della tabella

Il metodo `head()` restituisce le prime righe di un DataFrame. L’argomento `2` limita il risultato alle prime due.


````python
print(data.head(2))
````

L’attributo `shape` contiene le dimensioni della tabella nella forma `(numero di righe, numero di colonne)`.


````python
print(data.shape)
print(data.shape[0])
print(data.shape[1])
````

Il risultato è `(4, 3)`, seguito da `4` e `3`. La coppia di dimensioni è una **tupla**, una sequenza i cui elementi si leggono mediante indici. `data.shape[0]` restituisce il numero di righe e `data.shape[1]` il numero di colonne.

L’attributo `columns` contiene le etichette delle colonne. Il metodo `describe()` produce un riepilogo statistico delle colonne numeriche. Le righe `count`, `mean`, `min` e `max` indicano rispettivamente il numero di dati disponibili, la media, il minimo e il massimo.


````python
print(data.columns)
print(data.describe())
````

Il conteggio `count` vale `3` per `Area` e `4` per `Price`: la superficie è disponibile per tre delle quattro abitazioni. Le statistiche di ciascuna colonna vengono calcolate sui dati presenti.

### 10.2 Selezione delle colonne

Una colonna si seleziona specificandone il nome tra parentesi quadre.


````python
prices_column = data["Price"]
print(prices_column)
````

Il risultato contiene i quattro prezzi con le rispettive etichette di riga. La selezione di una singola colonna produce una **Series**, la struttura a una dimensione di pandas. La voce `dtype` visualizzata in fondo descrive il tipo dei dati contenuti.

Per selezionare più colonne si passa una lista dei loro nomi. Il risultato mantiene la struttura tabellare di un DataFrame.


````python
selected_names = ["Rooms", "Area"]
selected_data = data[selected_names]
print(selected_data)
````

La selezione può essere scritta anche nella forma `data[["Rooms", "Area"]]`. Le parentesi interne delimitano la lista dei nomi; quelle esterne applicano la selezione al DataFrame.

Una lista contenente un solo nome produce una tabella con una colonna. Il nome passato direttamente come stringa produce una Series.


````python
area_column = data["Area"]
area_table = data[["Area"]]
print(area_column.shape)
print(area_table.shape)
````

Le dimensioni sono `(4,)` per la Series e `(4, 1)` per il DataFrame: nel primo caso quattro valori in una dimensione, nel secondo quattro righe e una colonna. La virgola in `(4,)` identifica una tupla con un solo elemento.

### 10.3 Selezione delle righe

Il confronto di una colonna con un numero viene applicato a ogni suo elemento. L’espressione seguente verifica, per ciascuna abitazione, se la superficie è maggiore o uguale a 80 metri quadrati.


````python
area_filter = data["Area"] >= 80
print(area_filter)
````

La Series risultante contiene, nell’ordine, `False`, `True`, `True`, `False`. Le righe `1` e `2` soddisfano la condizione. Nella colonna numerica di questo esempio, il confronto con il valore mancante `NaN` restituisce `False`.

Una sequenza di valori booleani può essere usata come **filtro**: le righe associate a `True` vengono selezionate.


````python
large_houses = data[area_filter]
print(large_houses)
````

La tabella risultante contiene le abitazioni di superficie 80 e 120 metri quadrati. Le etichette di riga originali, `1` e `2`, vengono conservate.

Gli strumenti `.iloc` e `.loc` permettono ulteriori selezioni:

- `data.iloc[0]` restituisce la prima riga in base alla sua posizione;
- `data.loc[area_filter, ["Rooms", "Price"]]` applica il filtro alle righe e seleziona le colonne indicate per nome.

Nella forma `.loc[righe, colonne]`, la virgola separa i due criteri di selezione.


````python
print(data.iloc[0])
print(data.loc[area_filter, ["Rooms", "Price"]])
````

### 10.4 Valori mancanti

Il metodo `isna()` individua i dati mancanti. Per ogni elemento restituisce `True` se il valore manca, `False` se è presente. Applicato alla colonna `Area`, produce il seguente risultato.


````python
missing_area = data["Area"].isna()
print(missing_area)
print(missing_area.sum())
````

I risultati sono `False`, `False`, `False`, `True`. La somma vale `1`, perché nei calcoli booleani `True` contribuisce con 1 e `False` con 0.

La scrittura `data.isna().sum()` esegue due operazioni successive: individua i dati mancanti nella tabella e ne calcola il conteggio per colonna.

Il metodo `dropna()` permette di escludere le righe con dati mancanti. L’argomento `subset` specifica le colonne da esaminare. Il risultato viene assegnato a una nuova variabile, `clean_data`.


````python
print(data.isna().sum())
clean_data = data.dropna(subset=["Rooms", "Area", "Price"])
print(clean_data)
````

La tabella `clean_data` contiene tre righe. La quarta abitazione è stata esclusa perché il valore di `Area` è mancante. La variabile `data` conserva la tabella originale.

L’eliminazione delle righe è una possibile strategia di trattamento dei dati mancanti. La sua applicazione richiede di valutare quante osservazioni si perdono e quali conseguenze ne derivano per l’analisi.

### 10.5 Caratteristiche e valore da prevedere

Nel problema utilizzato come esempio, il prezzo deve essere previsto a partire da numero di stanze e superficie. Nei notebook del corso si adotta la seguente notazione:

- **`y`** indica la colonna del valore da prevedere, detto anche **target**: in questo caso `Price`.

La distinzione tra `X` maiuscola e `y` minuscola appartiene alla convenzione adottata negli esempi.

Le dimensioni di `X` sono `(3, 2)`: tre abitazioni descritte da due caratteristiche. Le dimensioni di `y` sono `(3,)`: un prezzo per ciascuna abitazione. Le righe delle due strutture corrispondono nello stesso ordine.

La colonna `Price` costituisce il risultato da prevedere e viene quindi tenuta separata dalle caratteristiche disponibili per una nuova abitazione.

## 11. Array NumPy

**NumPy** è una libreria per il calcolo numerico. La sua struttura principale è l’**array**, che organizza valori in una o più dimensioni e permette di eseguire operazioni numeriche sugli elementi.

L’abbreviazione convenzionale della libreria è `np`. La funzione `np.array()` costruisce un array a partire, per esempio, da una lista di numeri.


````python
import numpy as np

values = [10, 20, 30]
numbers = np.array(values)
print(numbers)
print(numbers * 2)
````

L’espressione `numbers * 2` produce `[20 40 60]`: ogni elemento dell’array viene moltiplicato per 2.

Per una lista Python, la stessa scrittura ha un altro significato: `values * 2` produce `[10, 20, 30, 10, 20, 30]`, ottenuto ripetendo due volte la sequenza. Il comportamento dell’operatore dipende dal tipo dei valori su cui opera.

### Array a due dimensioni

Un array a due dimensioni rappresenta una tabella di numeri, o **matrice**. Nella costruzione seguente ogni lista interna specifica una riga, con numero di stanze e superficie di un’abitazione.


````python
matrix = np.array([
    [2, 60],
    [3, 80],
    [4, 120],
])
print(matrix)
print(matrix.shape)
````

L’attributo `shape` vale `(3, 2)`: la matrice ha tre righe e due colonne.

La selezione degli elementi si scrive nella forma `[righe, colonne]`. Un indice individua una posizione; il simbolo `:` seleziona tutte le posizioni della dimensione corrispondente.


````python
print(matrix[0, 1])  # Riga 0, colonna 1: superficie della prima casa.
print(matrix[:, 0])  # Tutte le righe, colonna 0: numero di stanze.
````

L’espressione `matrix[0, 1]` restituisce `60`, elemento della prima riga e della seconda colonna. L’espressione `matrix[:, 0]` restituisce `[2 3 4]`, corrispondente a tutte le righe della prima colonna.

Nei notebook di Deep Learning, `X.shape[1]` indica il numero di caratteristiche: l’attributo `shape` fornisce le dimensioni e l’indice `1` seleziona il numero di colonne. Nell’esempio vale `2`.

## 13. Interpretazione degli errori

Quando un’istruzione provoca un errore, Python visualizza un messaggio con la posizione del problema e il tipo di errore rilevato. L’ultima riga riporta il nome dell’eccezione e una descrizione; le righe precedenti permettono di individuare l’istruzione coinvolta.

La verifica comprende la lettura del messaggio, il controllo della riga indicata e dei valori utilizzati, la correzione del codice e una nuova esecuzione della cella.

Per esempio, `print(prcie)` provoca `NameError` se la variabile definita in precedenza si chiama `price`. In questo caso la correzione consiste nel ripristinare il nome `price`.

| Errore | Causa tipica | Verifica |
| --- | --- | --- |
| `NameError` | Uso di un nome privo di definizione | Controllare l’ortografia del nome e l’esecuzione della cella che lo definisce. |
| `SyntaxError` | Scrittura incompatibile con le regole del linguaggio | Controllare parentesi, virgolette e due punti. |
| `IndentationError` | Rientro mancante o incoerente | Verificare l’allineamento delle istruzioni dei blocchi. |
| `IndexError` | Indice oltre le posizioni disponibili | Confrontare l’indice con la lunghezza della sequenza. |
| `KeyError` | Chiave o nome di colonna assente | Verificare la chiave e, per un DataFrame, le etichette in `data.columns`. |
| `TypeError` | Operazione applicata a tipi incompatibili | Controllare il tipo dei valori, per esempio nell’espressione `"80" + 20`. |
| `FileNotFoundError` | File assente nel percorso specificato | Verificare percorso, cartella di lavoro ed esecuzione della preparazione. |
| `ModuleNotFoundError` | Modulo assente dall’ambiente utilizzato | Verificare la preparazione e l’ambiente Python selezionato. |

Il frammento seguente illustra un errore di indentazione:

```text
if price < 200000:
print("Nel budget")
```

Il blocco associato a `if` richiede un rientro. La forma corretta è:

```python
if price < 200000:
    print("Nel budget")
```
