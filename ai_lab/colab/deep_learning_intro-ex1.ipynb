# Facoltativo: uscita di un modello non addestrato
Fino alla lezione 5 affronteremo regressioni: cerchiamo una funzione che si adatti a target numerici. Con un'unità lineare questa funzione è una retta.

I pesi iniziali sono casuali. Esegui più volte la cella seguente per vedere rette diverse. Non devi completare codice: è una dimostrazione.


````python
import tensorflow as tf
import matplotlib.pyplot as plt

model = keras.Sequential([
    layers.Dense(1, input_shape=[1]),
])

x = tf.linspace(-1.0, 1.0, 100)
y = model.predict(x)

plt.figure(dpi=100)
plt.plot(x, y, 'k')
plt.xlim(-1, 1)
plt.ylim(-1, 1)
plt.xlabel("Input: x")
plt.ylabel("Target y")
w, b = model.weights # Qui puoi usare anche model.get_weights().
plt.title("Weight: {:0.2f}\nBias: {:0.2f}".format(w[0][0], b[0]))
plt.show()
````
