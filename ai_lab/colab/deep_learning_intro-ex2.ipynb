# 3) Strati di attivazione

Esploriamo un po’ le funzioni di attivazione.

Il modo consueto per associare un’attivazione a uno strato `Dense` è includerla nella definizione con l’argomento `activation`. A volte, però, vorrai inserire un altro strato tra il `Dense` e la sua attivazione. Ne vedremo un esempio nella lezione 5, con la *batch normalization*. In questo caso possiamo definire l’attivazione in un proprio strato `Activation`:

```
layers.Dense(units=8),
layers.Activation('relu')
```

È del tutto equivalente alla forma ordinaria: `layers.Dense(units=8, activation='relu')`.

Riscrivi il modello seguente in modo che ciascuna attivazione si trovi in un proprio strato `Activation`.


````python
### SCRIVI QUI IL CODICE: riscrivi usando layer di attivazione.
model = keras.Sequential([
    layers.Dense(32, activation='relu', input_shape=[8]),
    layers.Dense(32, activation='relu'),
    layers.Dense(1),
])

# Controlla la risposta.
q_3.check()
````

# Facoltativo: alternative alla ReLU

Esiste un’intera famiglia di varianti dell’attivazione `'relu'`, tra cui `'elu'`, `'selu'` e `'swish'`, tutte utilizzabili in Keras. A volte un’attivazione funziona meglio di un’altra per un determinato compito: puoi quindi provare attivazioni diverse mentre sviluppi un modello. ReLU tende a funzionare bene nella maggior parte dei problemi, quindi è un buon punto di partenza.

Osserviamo i grafici di alcune di queste funzioni. Sostituisci `'relu'` con una delle altre attivazioni nominate sopra, poi esegui la cella per vedere il grafico. Consulta la [documentazione](https://www.tensorflow.org/api_docs/python/tf/keras/activations) per altre idee.


````python
# SCRIVI QUI IL CODICE: prova 'elu', 'selu', 'swish' o altre attivazioni al posto di 'relu'.
activation_layer = layers.Activation('relu')

x = tf.linspace(-3.0, 3.0, 100)
y = activation_layer(x) # Una volta creato, un layer si può chiamare come una funzione.

plt.figure(dpi=100)
plt.plot(x, y)
plt.xlim(-3, 3)
plt.xlabel("Input")
plt.ylabel("Output")
plt.show()
````
