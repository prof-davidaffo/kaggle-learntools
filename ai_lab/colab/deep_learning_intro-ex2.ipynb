# Facoltativo: alternative a ReLU
Keras offre `elu`, `selu`, `swish` e altre attivazioni. Su alcuni problemi una variante può funzionare meglio; ReLU resta un buon punto di partenza. Cambia il nome dell'attivazione nella cella seguente e osserva il grafico. Consulta la [documentazione](https://www.tensorflow.org/api_docs/python/tf/keras/activations) per altre opzioni.


````python
# SCRIVI QUI IL CODICE: prova 'elu', 'selu', 'swish' o altre attivazioni al posto di 'relu'.
activation_layer = layers.Activation('relu')

x = tf.linspace(-3.0, 3.0, 100)
y = activation_layer(x) # Una volta creato, un layer si può chiamare come una funzione.

plt.figure(dpi=100)
plt.plot(x, y)
plt.xlim(-3, 3)
plt.xlabel("Input")
plt.ylabel("Output")
plt.show()
````
