# Introduzione
Cominceremo con una panoramica di come funzionano i modelli di machine learning e di come vengono utilizzati. Se hai già lavorato con la modellazione statistica o il machine learning, potrebbe sembrarti elementare. Non preoccuparti: presto passeremo a costruire modelli potenti.

Durante questo corso costruirai modelli seguendo lo scenario descritto qui sotto.

Tuo cugino ha guadagnato milioni speculando sul mercato immobiliare. Visto il tuo interesse per la data science, ti ha proposto di diventare soci. Lui metterà il denaro e tu fornirai modelli capaci di prevedere quanto valgono diverse case.

Gli chiedi come abbia previsto finora il valore degli immobili e lui risponde che si tratta soltanto di intuito. Facendogli altre domande, però, scopri che ha individuato schemi nei prezzi delle case viste in passato e che usa questi schemi per fare previsioni sulle nuove case che sta prendendo in considerazione.

Il machine learning funziona allo stesso modo. Partiremo da un modello chiamato **albero decisionale**, o *Decision Tree*. Esistono modelli più sofisticati che producono previsioni più accurate. Ma gli alberi decisionali sono facili da comprendere e costituiscono l’elemento di base di alcuni dei migliori modelli della data science.

Per semplicità, cominceremo dall’albero decisionale più semplice possibile.

![Primi alberi decisionali](https://storage.googleapis.com/kaggle-media/learn/images/7tsb5b1.png)

Divide le case in due sole categorie. Il prezzo previsto per qualsiasi casa presa in esame è la media storica dei prezzi delle case appartenenti alla stessa categoria.

Usiamo i dati per decidere come dividere le case nei due gruppi e, poi, per determinare il prezzo previsto in ciascun gruppo. Questa fase, in cui si ricavano schemi dai dati, si chiama **fitting** o **training**, cioè addestramento del modello. I dati usati per **addestrare** il modello si chiamano **dati di training**.

I dettagli dell’addestramento, per esempio come suddividere i dati, sono abbastanza complessi da meritare una spiegazione successiva. Dopo aver addestrato il modello, puoi applicarlo a nuovi dati per **prevedere** i prezzi di altre case.

# Migliorare l’albero decisionale
Quale dei due alberi seguenti è più probabile che si ottenga addestrando il modello sui dati del mercato immobiliare?

![Primi alberi decisionali](https://storage.googleapis.com/kaggle-media/learn/images/prAjgku.png)

L’albero di sinistra, Decision Tree 1, probabilmente ha più senso: rappresenta il fatto che le case con più camere da letto tendono a essere vendute a prezzi più alti di quelle con meno camere. Il limite principale di questo modello è che non tiene conto della maggior parte dei fattori che influenzano il prezzo, come il numero di bagni, la superficie del terreno, la posizione e così via.

Puoi considerare più fattori usando un albero con un numero maggiore di suddivisioni, dette *split*. Questi alberi vengono chiamati «più profondi». Un albero che consideri anche la superficie totale del terreno di ogni casa potrebbe avere questo aspetto:

![Albero di profondità 2](https://storage.googleapis.com/kaggle-media/learn/images/R3ywQsR.png)

Per prevedere il prezzo di una casa, percorri l’albero scegliendo sempre il ramo corrispondente alle sue caratteristiche. Il prezzo previsto si trova in fondo all’albero. Il punto finale in cui viene formulata una previsione si chiama **foglia**, o *leaf*.

Le suddivisioni e i valori delle foglie saranno determinati dai dati: è quindi il momento di esaminare i dati con cui lavorerai.
