# Introduzione
Cominciamo da come funzionano i modelli di machine learning e da come si usano. Se hai già esperienza di statistica o machine learning, alcuni passaggi ti sembreranno semplici: presto costruiremo modelli più potenti.

Immagina questa situazione: tuo cugino ha guadagnato molto nel mercato immobiliare e ti propone di collaborare. Lui mette il capitale, tu costruisci modelli per prevedere il valore delle case. Dice di basarsi sull'intuito, ma interrogandolo scopri che riconosce schemi nei prezzi delle case già viste e li applica ai nuovi immobili.

Il machine learning funziona in modo simile. Partiamo dal **decision tree**, l'albero decisionale. Esistono modelli più sofisticati, ma gli alberi sono comprensibili e costituiscono la base di tecniche molto efficaci.

![Primo albero decisionale](https://storage.googleapis.com/kaggle-media/learn/images/7tsb5b1.png)

L'albero più semplice divide le case in due categorie. Per una casa nuova, il prezzo previsto è il prezzo medio storico delle case della sua categoria. I dati determinano sia la divisione in gruppi sia il prezzo assegnato a ogni gruppo. Apprendere questi schemi significa **addestrare il modello**, in inglese *training* o *fitting*. I dati impiegati sono i **training data**.

Rimandiamo i dettagli dell'algoritmo che sceglie le divisioni. Una volta addestrato, il modello può prevedere il prezzo di case nuove.

# Migliorare l'albero
Quale dei seguenti alberi potrebbe derivare più probabilmente da dati immobiliari reali?

![Due alberi a confronto](https://storage.googleapis.com/kaggle-media/learn/images/prAjgku.png)

Quello a sinistra è più plausibile: le case con più camere tendono a costare di più. Tuttavia ignora bagni, superficie del terreno, posizione e molti altri fattori.

Possiamo considerare più caratteristiche con un albero **più profondo**, che effettua più divisioni, dette *split*. Aggiungendo la superficie del terreno potremmo ottenere questo albero:

![Albero di profondità due](https://storage.googleapis.com/kaggle-media/learn/images/R3ywQsR.png)

Per fare una previsione segui il percorso corrispondente alle caratteristiche della casa. Il nodo finale, che contiene la previsione, è una **foglia**, o *leaf*. Sia le divisioni sia i valori nelle foglie dipendono dai dati: ora è il momento di esaminarli.
