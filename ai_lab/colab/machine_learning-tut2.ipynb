# Interpretare le statistiche descrittive
Il risultato mostra otto statistiche per ogni colonna numerica. **count** conta le righe con un valore presente. I valori mancanti possono avere varie cause: per una casa con una sola camera, per esempio, la superficie della seconda camera non esiste. Torneremo sul problema.

**mean** è la media; **std** è la deviazione standard e misura quanto i valori sono dispersi.

Per capire **min**, **25%**, **50%**, **75%** e **max**, immagina di ordinare i valori dal più piccolo al più grande. `min` e `max` sono gli estremi. Il **25° percentile** si trova a un quarto della distribuzione: circa il 25% dei valori è inferiore e il 75% è superiore. Analogamente si interpretano il 50° percentile, cioè la mediana, e il 75° percentile.
