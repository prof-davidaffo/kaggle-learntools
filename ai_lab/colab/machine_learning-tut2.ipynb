# Interpretare la descrizione dei dati
Il risultato mostra otto numeri per ciascuna colonna del dataset originale. Il primo, **count**, indica quante righe contengono valori non mancanti.

I valori possono mancare per molte ragioni. Per esempio, rilevando le caratteristiche di una casa con una sola camera da letto, non si raccoglie la superficie della seconda camera. Torneremo sul tema dei dati mancanti.

Il secondo valore, **mean**, è la media. Subito sotto, **std** è la deviazione standard, che misura quanto i valori siano numericamente dispersi.

Per interpretare **min**, **25%**, **50%**, **75%** e **max**, immagina di ordinare ogni colonna dal valore più basso a quello più alto. Il primo valore, il più piccolo, è `min`. Percorrendo un quarto dell’elenco, trovi un numero più grande del 25% dei valori e più piccolo del restante 75%. È il valore **25%**, che si legge «venticinquesimo percentile». Il cinquantesimo e il settantacinquesimo percentile si definiscono in modo analogo, mentre **max** è il numero più grande.
