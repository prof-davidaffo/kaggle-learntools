## 2. Assegnazioni e operatori

In Python l’assegnazione associa un nome a un oggetto e il tipo viene determinato dal valore assegnato. Il tipo appartiene al valore e lo stesso nome può essere riassegnato anche a un valore di tipo diverso. Negli esempi `price` è il prezzo e `area` la superficie.


````python
price = 200000
area = 80
price_per_square_meter = price / area

print(price)
print(price_per_square_meter)
````

Le stampe danno `200000` e `2500.0`. L’operatore `/` restituisce un `float` anche quando la divisione tra interi è esatta. I nomi distinguono maiuscole e minuscole; per separare le parole si usa normalmente `_`, come in `price_per_square_meter`.

Una nuova assegnazione fa riferimento al nuovo valore:


````python
rooms = 3
print(rooms)
rooms = 4
print(rooms)
````

Le stampe danno `3` e `4`. Ecco gli operatori numerici più utili nel corso:

| Operatore | Significato | Esempio |
| --- | --- | --- |
| `+`, `-`, `*` | Somma, differenza, prodotto | `10 * 3` → `30` |
| `/` | Divisione | `10 / 4` → `2.5` |
| `//` | Divisione con arrotondamento verso il basso | `10 // 4` → `2` |
| `%` | Resto | `10 % 4` → `2` |
| `**` | Potenza | `3 ** 2` → `9` |

I decimali si scrivono con il punto. Come di consueto, le parentesi modificano la precedenza degli operatori.

## 3. Tipi e confronti

Tra i tipi predefiniti incontrerai spesso `int` (interi), `float` (numeri in virgola mobile), `str` (stringhe) e `bool` (`True` o `False`). Le virgolette fanno la differenza: `80` è un intero, `"80"` una stringa.


````python
print(80 + 20)
print("80" + "20")
````

Il primo risultato è `100`; il secondo è `8020`, perché `+` concatena le stringhe. Puoi controllare il tipo con `type(80)`. Per convertire esplicitamente una stringa numerica usa, per esempio, `int("80")` o `float("80")`.

I confronti usano `==`, `!=`, `<`, `<=`, `>` e `>=` e restituiscono `True` o `False`. Come in altri linguaggi, `=` assegna e `==` confronta. Gli operatori logici si scrivono `and`, `or` e `not`.


````python
price = 180000
budget = 200000
within_budget = price <= budget

print(within_budget)
````

Qui `within_budget` vale `True`. Prova a portare `price` a `250000`: il confronto restituirà `False`.


## 4. Condizioni e indentazione

In Python l’indentazione delimita i blocchi di istruzioni. Dopo la condizione si scrivono i due punti; per convenzione ogni livello rientra di quattro spazi.


````python
price = 180000
budget = 200000

if price <= budget:
    print("La casa rientra nel budget")
else:
    print("La casa supera il budget")

print("Controllo terminato")
````

Con questi valori vengono stampati `La casa rientra nel budget` e `Controllo terminato`. I due `print()` rientrati appartengono ai rispettivi rami; l’ultimo è fuori dalla condizione e viene eseguito in entrambi i casi. Se porti `price` a `250000`, viene eseguito il ramo `else`.

Per aggiungere un caso intermedio usa `elif`, seguito dalla condizione e dai due punti. Puoi scrivere la condizione direttamente, come nell’esempio:

```python
if price < budget:
    print("Sotto il budget")
elif price == budget:
    print("Esattamente nel budget")
else:
    print("Sopra il budget")
```

## 5. Liste e accesso agli elementi

Le liste Python sono sequenze ordinate e modificabili. Si scrivono tra parentesi quadre; `len()` ne restituisce la lunghezza.


````python
prices = [150000, 200000, 270000]
print(prices)
print(len(prices))
````

La prima stampa mostra la lista, la seconda `3`. Gli indici partono da zero: i tre prezzi hanno quindi indici `0`, `1` e `2`.


````python
print(prices[0])  # Primo prezzo: 150000.
print(prices[1])  # Secondo prezzo: 200000.
print(prices[2])  # Terzo prezzo: 270000.
````

Un indice fuori dai limiti, come `prices[3]`, provoca `IndexError`. Gli indici negativi partono dalla fine: `prices[-1]` è l’ultimo elemento.

### Selezione di un intervallo

Lo *slicing* `[inizio:fine]` restituisce una nuova lista: l’inizio è incluso e la fine esclusa. Per esempio, `prices[0:2]` comprende gli elementi agli indici `0` e `1`.


````python
first_two_prices = prices[0:2]
print(first_two_prices)
````

Il risultato è `[150000, 200000]`.

L’omissione dell’indice iniziale estende la selezione all’inizio della lista; l’omissione dell’indice finale la estende fino alla fine.

| Espressione | Intervallo selezionato | Risultato |
| --- | --- | --- |
| `prices[:2]` | Dall’inizio fino all’indice 2 escluso | `[150000, 200000]` |
| `prices[1:]` | Dall’indice 1 incluso fino alla fine | `[200000, 270000]` |
| `prices[:]` | Tutti gli elementi, in una nuova lista | `[150000, 200000, 270000]` |

## 6. Ciclo `for`

Il `for` Python scorre direttamente gli elementi di un oggetto iterabile. Qui `price` assume a turno ogni valore della lista; i due punti e il rientro delimitano il corpo del ciclo.


````python
prices = [150000, 250000, 180000]

for price in prices:
    print("Prezzo:", price)

print("Elenco terminato")
````

La cella stampa i tre prezzi, poi `Elenco terminato` una sola volta, fuori dal ciclo. `print()` separa con uno spazio gli argomenti ricevuti.

Quando servono indici numerici puoi usare `range(3)`, che produce `0`, `1`, `2`; per ottenere insieme indice e valore usa `enumerate(prices)`.

### Blocchi annidati

Ogni blocco annidato aggiunge un livello di indentazione. Qui il controllo del budget viene ripetuto per ogni prezzo.


````python
for price in prices:
    if price <= 200000:
        print("Nel budget:", price)

print("Controlli terminati")
````

Risultato:

```text
Nel budget: 150000
Nel budget: 180000
Controlli terminati
```

L’istruzione `if` appartiene al blocco del `for` e ha un rientro di quattro spazi. Il `print()` condizionale appartiene al blocco dell’`if` e ha un rientro di otto spazi. La stampa finale è esterna a entrambi i blocchi.

## 7. Funzioni

Le funzioni predefinite comprendono `print()`, `sum()` e `len()`. Per esempio, possiamo usare le ultime due per calcolare una media:


````python
values = [10, 20, 30]
total = sum(values)
count = len(values)
print(total)
print(count)
print(total / count)
````

Le stampe danno `60`, `3` e `20.0`. Una funzione si definisce con `def`, seguito dai parametri, dai due punti e dal corpo rientrato:


````python
def unit_cost(price, area):
    return price / area
````

`return` restituisce il risultato al chiamante. La definizione crea la funzione; il corpo viene eseguito quando la funzione è chiamata.


````python
first_cost = unit_cost(200000, 80)
second_cost = unit_cost(270000, 120)
print(first_cost)
print(second_cost)
````

Le chiamate restituiscono rispettivamente `2500.0` e `2250.0`. Gli argomenti possono essere passati per posizione o per nome, una forma frequente nelle librerie del corso:


````python
print(unit_cost(200000, 80))
print(unit_cost(price=200000, area=80))
````

Le due chiamate sono equivalenti. Una funzione senza `return` esplicito restituisce `None`. Anche `print()` restituisce `None`: il suo scopo è visualizzare i valori, mentre `return` consegna un risultato al chiamante.

## 8. Dizionari

Un **dizionario** è una raccolta di associazioni tra **chiavi** e **valori**. L’accesso a un valore avviene mediante la chiave corrispondente.

Nell’esempio, le chiavi `"Rooms"`, `"Area"` e `"Price"` identificano le caratteristiche di una casa. Il dizionario è delimitato da parentesi graffe; i due punti separano ciascuna chiave dal valore associato.


````python
house = {"Rooms": 3, "Area": 80.0, "Price": 200000}
print(house["Area"])

house["Rooms"] = 4
print(house["Rooms"])
````

Le stampe producono `80.0` e `4`. L’espressione `house["Area"]` restituisce il valore associato alla chiave `"Area"`. L’assegnazione a `house["Rooms"]` aggiorna il numero di stanze.

Le chiavi usate nell’esempio sono stringhe e vengono scritte tra virgolette. La distinzione tra maiuscole e minuscole vale anche in questo caso: `"Area"` e `"area"` sono chiavi diverse.

### Riepilogo delle parentesi

| Parentesi | Esempio | Uso |
| --- | --- | --- |
| Tonde `()` | `print(price)` | Chiamata di una funzione |
| Tonde `()` | `(price + 10000) / area` | Raggruppamento di un’espressione |
| Quadre `[]` | `[150000, 200000]` | Creazione di una lista |
| Quadre `[]` | `prices[0]`, `house["Area"]` | Accesso a un elemento |
| Graffe `{}` | `{"Rooms": 3}` | Creazione di un dizionario |


## 9. Importazioni, attributi e metodi

L’istruzione `import` rende disponibile un modulo. Per esempio, `math` fa parte della libreria standard di Python e contiene funzioni matematiche:


````python
import math

print(math.sqrt(25))
````

Il punto permette di accedere ai nomi del modulo: `math.sqrt(25)` restituisce `5.0`. Con `import math as m` puoi scegliere un nome locale più breve e scrivere `m.sqrt(25)`.

La stessa notazione si usa per attributi e metodi degli oggetti. Per esempio, `prices.append(300000)` aggiunge un elemento alla lista `prices`. Le parentesi indicano la chiamata del metodo.


## 10. Interpretazione degli errori

Quando un’istruzione provoca un errore, Python visualizza un messaggio con la posizione del problema e il tipo di errore rilevato. L’ultima riga riporta il nome dell’eccezione e una descrizione; le righe precedenti permettono di individuare l’istruzione coinvolta.

La verifica comprende la lettura del messaggio, il controllo della riga indicata e dei valori utilizzati, la correzione del codice e una nuova esecuzione della cella.

Per esempio, `print(prcie)` provoca `NameError` se la variabile definita in precedenza si chiama `price`. In questo caso la correzione consiste nel ripristinare il nome `price`.

| Errore | Causa tipica | Verifica |
| --- | --- | --- |
| `NameError` | Uso di un nome privo di definizione | Controllare l’ortografia del nome e l’esecuzione della cella che lo definisce. |
| `SyntaxError` | Scrittura incompatibile con le regole del linguaggio | Controllare parentesi, virgolette e due punti. |
| `IndentationError` | Rientro mancante o incoerente | Verificare l’allineamento delle istruzioni dei blocchi. |
| `IndexError` | Indice oltre le posizioni disponibili | Confrontare l’indice con la lunghezza della sequenza. |
| `KeyError` | Chiave assente dal dizionario | Verificare le chiavi disponibili nel dizionario. |
| `TypeError` | Operazione applicata a tipi incompatibili | Controllare il tipo dei valori, per esempio nell’espressione `"80" + 20`. |
| `FileNotFoundError` | File assente nel percorso specificato | Verificare percorso, cartella di lavoro ed esecuzione della preparazione. |
| `ModuleNotFoundError` | Modulo assente dall’ambiente utilizzato | Verificare la preparazione e l’ambiente Python selezionato. |

Il frammento seguente illustra un errore di indentazione:

```text
if price < 200000:
print("Nel budget")
```

Il blocco associato a `if` richiede un rientro. La forma corretta è:

```python
if price < 200000:
    print("Nel budget")
```
