## 2. Valori, variabili e operatori

In Python non scrivi una dichiarazione di tipo prima di usare una variabile. Con `price = 200000` crei il nome `price` e lo associ al valore intero `200000`. Il simbolo `=` indica un’**assegnazione**: Python valuta prima l’espressione a destra, poi associa il risultato al nome a sinistra.

Un nome deve essere definito prima di poterlo leggere: il primo uso di un nome sconosciuto non gli assegna automaticamente zero. Negli esempi usiamo nomi inglesi, come nei notebook del corso: `price` significa prezzo, `area` superficie. I nomi distinguono maiuscole e minuscole: `price` e `Price` sono due nomi diversi. Non possono contenere spazi; per separare le parole usiamo `_`.

Nella cella seguente definiamo prima i due dati, poi li usiamo in un calcolo.


````python
price = 200000
area = 80
price_per_square_meter = price / area

print(price)
print(price_per_square_meter)
````

Le stampe danno `200000` e `2500.0`. Quando legge `price / area`, Python usa i valori associati a quei nomi, quindi calcola `200000 / 80`. Il risultato viene memorizzato con il nome `price_per_square_meter` (prezzo al metro quadrato).

La divisione `/` produce un numero con parte decimale anche se il risultato matematico è intero. In Python i decimali si scrivono con il punto: `2500.0`.

Una nuova assegnazione sostituisce il valore associato a un nome:


````python
rooms = 3
print(rooms)
rooms = 4
print(rooms)
````

Le stampe danno `3` e poi `4`. Possiamo anche aggiornare un valore a partire da quello precedente: `rooms = rooms + 1` prima calcola `4 + 1`, poi assegna `5` a `rooms`.

**Un calcolo assegnato non è una formula che si aggiorna da sola.** Se cambi `price`, il valore già calcolato di `price_per_square_meter` resta quello precedente finché non esegui nuovamente l’assegnazione con la divisione.

| Operatore | Significato | Esempio e risultato |
| --- | --- | --- |
| `+`, `-`, `*` | Somma, differenza, prodotto | `10 * 3` → `30` |
| `/` | Divisione | `10 / 4` → `2.5` |
| `//` | Divisione arrotondata verso il basso | `10 // 4` → `2` |
| `%` | Resto | `10 % 4` → `2` |
| `**` | Potenza | `3 ** 2` → `9` |

Se provieni da un linguaggio in cui la divisione tra interi elimina la parte decimale, fai attenzione: `5 / 2` in Python vale `2.5`. `//` arrotonda verso il basso, anche con numeri negativi: `-5 // 2` vale `-3`, non `-2`. Non si limita quindi a eliminare la parte decimale. Per elevare a potenza si usa `**`: `^` non è l’operatore di potenza.

Prodotto e divisione precedono somma e sottrazione: `2 + 3 * 4` vale `14`, mentre `(2 + 3) * 4` vale `20`. Le parentesi rendono esplicito l’ordine desiderato.

## 3. Tipi, conversioni e confronti

Il tipo appartiene al **valore**, non al nome della variabile: puoi scrivere prima `value = 80` e poi `value = "ottanta"`. Python accetta la riassegnazione, ma le operazioni successive dipendono dal nuovo tipo. Per esempio, `value / 2` funziona nel primo caso e produce `TypeError` nel secondo.

La tipizzazione dinamica non significa che qualunque operazione sia consentita. Ecco i tipi che useremo:

| Tipo | Che cosa rappresenta | Esempi |
| --- | --- | --- |
| `int` | Numeri interi | `80`, `-3`, `0` |
| `float` | Numeri in virgola mobile | `80.0`, `2.5` |
| `str` | Stringhe, cioè sequenze di caratteri | `"Roma"`, `"80"` |
| `bool` | Valori di verità | `True`, `False` |

Le stringhe possono usare apici singoli o doppi, purché l’apertura e la chiusura coincidano. Le virgolette non fanno parte del testo: indicano dove inizia e finisce. `80` è un numero; `"80"` è una stringa di due caratteri.

Osserva come cambia il significato di `+`:


````python
print(80 + 20)
print("80" + "20")
````

Il primo risultato è `100`; il secondo è `8020`: tra numeri `+` esegue la somma, tra stringhe unisce i testi (**concatenazione**). L’espressione `"80" + 20` provoca invece un errore: Python non sceglie automaticamente se sommare o unire.

Per conoscere il tipo di un valore usa `type(...)`; per convertirlo usa `int(...)`, `float(...)` o `str(...)`:

````python
print(type("80"))       # <class 'str'>
print(int("80") + 20)   # 100
print(float("2.5"))     # 2.5
print("Stanze: " + str(3))  # Stanze: 3
````

La conversione deve avere senso: `int("ciao")` non può produrre un intero e genera `ValueError`. `int(2.9)` tronca la parte decimale e dà `2`: non arrotonda al numero più vicino. Anche `int(-2.5)` tronca verso zero e dà `-2`, a differenza di `-5 // 2`, che arrotonda verso il basso e dà `-3`.

### Se acquisisci un dato da tastiera

Nei notebook del corso useremo soprattutto valori assegnati nelle celle e, più avanti, dati letti da file. Se vuoi adattare un programma che conosci con input da tastiera, la funzione Python è `input(...)`: mostra un messaggio, attende il testo e lo restituisce come **stringa**, anche quando digiti cifre.

Puoi provare questo frammento in una nuova cella:

### Confrontare i valori

Un confronto produce un valore booleano: `True` (vero) oppure `False` (falso), con l’iniziale maiuscola e senza virgolette.

| Operatore | Domanda | Esempio |
| --- | --- | --- |
| `==` | Sono uguali? | `3 == 3` → `True` |
| `!=` | Sono diversi? | `3 != 3` → `False` |
| `<`, `>` | È minore / maggiore? | `3 > 5` → `False` |
| `<=`, `>=` | È minore o uguale / maggiore o uguale? | `3 <= 3` → `True` |

Attenzione alla differenza: `price = 180000` assegna un valore; `price == 180000` verifica se il valore è uguale a 180000.

````python
price = 180000
budget = 200000
within_budget = price <= budget

print(within_budget)
````

Qui `within_budget` vale `True`: anche il risultato di un confronto può essere conservato in una variabile. Prova a portare `price` a `250000` e riesegui la cella: il risultato sarà `False`.

Gli operatori logici che conosci si scrivono `and`, `or` e `not`. Per combinare confronti non usare `&&`, `||`, `!` o gli operatori `&` e `|`:

- `and` dà `True` se entrambe le condizioni sono vere;
- `or` dà `True` se almeno una è vera, anche quando lo sono entrambe;
- `not` inverte il valore di verità.

````python
price = 180000
area = 80
print(price <= 200000 and area >= 70)  # True: entrambe vere
print(price <= 200000 or area >= 100)  # True: basta la prima
print(not (price <= 200000))           # False
````

**Prova tu:** cambia soltanto `area` in `60`. Prevedi i tre risultati prima di eseguire: saranno `False`, `True`, `False`.

## 4. Condizioni e indentazione

`if` ed `else` esprimono la selezione che già conosci. In Python la condizione non richiede parentesi tonde e i rami non sono racchiusi tra graffe. La riga con `if` termina con **due punti**; anche `else` termina con due punti, ma non ha una condizione.

In Python i gruppi di istruzioni, detti **blocchi**, sono delimitati dai rientri a inizio riga (**indentazione**). Il rientro fa parte della sintassi: stabilisce quali istruzioni dipendono dall’`if`. Non si usano parentesi graffe per delimitare i blocchi.

### Come scrivere i rientri senza contare gli spazi

**Usa il tasto Tab per entrare in un blocco.** Negli editor dei notebook il rientro è normalmente automatico dopo una riga terminata da `:`. Se la nuova riga è già rientrata, non aggiungere un altro Tab. Per tornare al livello precedente usa **Maiusc + Tab** all’inizio della riga; puoi anche selezionare più righe e rientrarle insieme.

Il tasto Tab e il carattere di tabulazione salvato nel file sono due cose diverse: un editor configurato per Python può inserire spazi quando premi Tab. I quattro spazi che vedi negli esempi rappresentano un livello di rientro, **non quattro pressioni della barra spaziatrice**. Python ammette anche tabulazioni, ma mischiarle con gli spazi può causare errori: lascia che l’editor mantenga uno stile coerente.

Leggi l’esempio osservando l’allineamento: `if` ed `else` sono allo stesso livello; le istruzioni di ciascun ramo sono un livello più all’interno.


````python
price = 180000
budget = 200000

if price <= budget:
    print("La casa rientra nel budget")
else:
    print("La casa supera il budget")

print("Controllo terminato")
````

Con questi valori vengono stampati `La casa rientra nel budget` e `Controllo terminato`. Python verifica `price <= budget`: essendo vero, esegue il primo blocco e salta il secondo. La stampa finale è tornata al margine sinistro, quindi viene eseguita dopo la scelta, in entrambi i casi.

Prova con `price = 250000`: verrà eseguito il ramo `else`. Prova poi con `price = 200000`: `<=` include l’uguaglianza, quindi la casa rientra nel budget.

### Scegliere tra più casi con `elif`

`elif` significa «altrimenti, se». Python esamina le condizioni dall’alto verso il basso ed esegue soltanto il primo ramo vero. Se nessuna è vera, esegue `else`, quando presente.

````python
if price < budget:
    print("Sotto il budget")
elif price == budget:
    print("Esattamente nel budget")
else:
    print("Sopra il budget")
````

Qui i tre casi distinguono prezzo inferiore, uguale e superiore al budget. Con due `if` separati, invece, Python esaminerebbe due scelte indipendenti. `else` è facoltativo: se manca e la condizione è falsa, il programma prosegue dopo il blocco.

### Condizioni composte: scrivi entrambi i confronti

Per verificare che il numero di stanze sia 2 oppure 3, scrivi `rooms == 2 or rooms == 3`. La forma `rooms == 2 or 3` non confronta `rooms` con 3: il secondo operando è il numero 3, che in una condizione viene considerato vero. Finiresti quindi per accettare anche un numero di stanze diverso da 2 e 3.

In una condizione Python considera falsi `False`, `None`, lo zero e le stringhe vuote; considera veri i numeri diversi da zero e le stringhe non vuote. Il testo `"False"`, per esempio, è una stringa non vuota, non il booleano `False`.

`and` e `or` valutano da sinistra a destra e si fermano appena il risultato è determinato. Questo comportamento, detto **cortocircuito**, permette di controllare un dato prima di usarlo:

````python
area = 0
price = 180000
if area > 0 and price / area <= 2500:
    print("Prezzo al metro quadrato accettabile")
else:
    print("Superficie non valida o prezzo troppo alto")
````

Con `area = 0`, il primo confronto è falso e la divisione non viene eseguita. Scambiando l’ordine dei confronti otterresti invece una divisione per zero. Prova anche `area = 80`: il primo confronto è vero, quindi viene valutato anche il secondo.

## 5. Liste e accesso agli elementi

Se hai già usato vettori o array, riconoscerai l’accesso per indice. La **lista** Python è una sequenza modificabile la cui lunghezza può cambiare durante l’esecuzione: non devi dichiarare in anticipo quanti elementi conterrà. Può contenere valori di tipi diversi, anche se qui raggrupperemo dati omogenei per rendere chiari i calcoli. Gli elementi sono scritti tra parentesi quadre e separati da virgole.

`prices` contiene tre prezzi. `print(prices)` mostra l’intera lista; `len(prices)` restituisce il numero dei suoi elementi. `len` è una funzione già disponibile in Python, come `print`. La lista vuota si scrive `[]` e ha lunghezza zero.


````python
prices = [150000, 200000, 270000]
print(prices)
print(len(prices))
````

La prima stampa mostra la lista, la seconda `3`. Gli indici partono da zero: i tre prezzi hanno quindi indici `0`, `1` e `2`.


````python
print(prices[0])  # Primo prezzo: 150000.
print(prices[1])  # Secondo prezzo: 200000.
print(prices[2])  # Terzo prezzo: 270000.
````

La posizione di un elemento si chiama **indice**. Per una lista di tre elementi gli indici validi sono `0`, `1`, `2`: l’ultimo indice è la lunghezza meno uno. `prices[3]` provoca quindi `IndexError`. `prices[-1]` è un modo comodo per ottenere l’ultimo elemento; `prices[-2]` indica il penultimo.

### Selezionare un intervallo

Lo **slicing** `[inizio:fine]` restituisce una nuova lista con una parte degli elementi: l’inizio è incluso, la fine è esclusa. `prices[0:2]` prende quindi gli indici `0` e `1`, fermandosi prima del `2`.

La differenza è essenziale: `prices[0]` restituisce un singolo prezzo; `prices[0:2]` restituisce una lista di due prezzi.


````python
first_two_prices = prices[0:2]
print(first_two_prices)
````

Il risultato è `[150000, 200000]`. Per prendere i primi due elementi possiamo anche scrivere `prices[:2]`: omettere l’inizio significa partire dal primo elemento. Omettere la fine significa arrivare in fondo.

| Espressione | Selezione | Risultato |
| --- | --- | --- |
| `prices[:2]` | Primi due elementi | `[150000, 200000]` |
| `prices[1:]` | Dal secondo elemento alla fine | `[200000, 270000]` |
| `prices[:]` | Tutti gli elementi, in una nuova lista | `[150000, 200000, 270000]` |

### Modificare una lista: assegnazione e metodi

Una lista è modificabile: `prices[1] = 210000` sostituisce il secondo elemento. Per aggiungere un elemento in fondo possiamo usare `prices.append(300000)`.

`append` è un **metodo**, cioè un’operazione legata a un oggetto. Si scrive dopo il nome della lista e un punto; tra parentesi passiamo il valore da aggiungere. Confronta `len(prices)`, una funzione che riceve la lista, con `prices.append(300000)`, un metodo chiamato su quella lista.


````python
prices = [150000, 200000, 270000]
prices[1] = 210000
prices.append(300000)
print(prices)
print(len(prices))
````

Otteniamo `[150000, 210000, 270000, 300000]` e `4`. `append` modifica la lista esistente e restituisce `None`: scrivi `prices.append(300000)`, senza assegnarne il risultato a `prices`.

### Assegnare una lista non significa copiarla

`other_prices = prices` crea un altro nome per la **stessa lista**. `prices[:]` crea invece una nuova lista. Questa differenza diventa visibile quando modifichi un elemento: prevedi l’output della cella seguente prima di eseguirla.


````python
prices = [150000, 200000]
other_prices = prices
copied_prices = prices[:]
other_prices[0] = 160000
print(prices)
print(copied_prices)
````

Le stampe sono `[160000, 200000]` e `[150000, 200000]`. La modifica tramite `other_prices` è visibile anche tramite `prices`, perché i due nomi indicano la stessa lista. `copied_prices` indica una lista diversa.

La copia con `[:]` è **superficiale**: crea una nuova lista, ma non copia ricorsivamente eventuali liste contenute al suo interno. Per gli esempi di questo capitolo, che contengono numeri o stringhe, è sufficiente. Anche una lista vuota, come la stringa vuota vista prima, è considerata falsa in una condizione; una lista con elementi è considerata vera.


## 6. Cicli: `for`, `range` e `while`

Il `for` Python scorre gli elementi di una sequenza: corrisponde a «per ogni elemento», anziché a un’intestazione con inizializzazione, condizione e incremento del contatore. Nella riga `for price in prices:`, `prices` è la lista da percorrere e `price` è il nome che riceve, uno alla volta, i suoi elementi. Il nome al singolare è una nostra scelta, non una regola del linguaggio.

A ogni passaggio Python assegna a `price` il valore successivo ed esegue tutto il blocco rientrato. Quando gli elementi finiscono, prosegue con le istruzioni esterne al ciclo. Non occorre aggiornare manualmente un indice.


````python
prices = [150000, 250000, 180000]

for price in prices:
    print("Prezzo:", price)

print("Elenco terminato")
````

La cella stampa i tre prezzi, poi `Elenco terminato` una sola volta. `print("Prezzo:", price)` riceve due argomenti separati da una virgola e li visualizza separandoli con uno spazio.

| Passaggio | Valore di `price` | Riga stampata |
| --- | --- | --- |
| 1 | `150000` | `Prezzo: 150000` |
| 2 | `250000` | `Prezzo: 250000` |
| 3 | `180000` | `Prezzo: 180000` |

La lista `prices` non viene modificata dal ciclo. Se fosse vuota, il corpo non verrebbe eseguito neppure una volta, ma la stampa finale comparirebbe comunque.

### Blocchi annidati

Possiamo inserire un `if` dentro un `for`: per ogni prezzo, controlliamo se rientra nel budget. Ogni blocco interno aggiunge un livello di rientro: il `for` è al margine, l’`if` un livello dentro, la stampa condizionata due livelli dentro.


````python
for price in prices:
    if price <= 200000:
        print("Nel budget:", price)

print("Controlli terminati")
````

Risultato:

```text
Nel budget: 150000
Nel budget: 180000
Controlli terminati
```

L’`if` viene eseguito tre volte, una per prezzo. Il suo `print` viene eseguito solo due volte, perché `250000 <= 200000` è falso. La stampa finale è esterna a entrambi i blocchi.

### Accumulare un risultato durante il ciclo

Per sommare i valori uno alla volta, inizializziamo un totale prima del ciclo e lo aggiorniamo a ogni passaggio. La posizione dell’inizializzazione è importante: dentro il ciclo azzererebbe il totale ogni volta.


````python
values = [10, 20, 30]
total = 0
for value in values:
    total = total + value
print(total)
````

Il totale passa da `0` a `10`, poi a `30`, infine a `60`. `total = total + value` usa il totale precedente per calcolare quello nuovo. Puoi incontrare la forma abbreviata `total += value`, che qui ha lo stesso effetto. Python non usa `total++` per incrementare un contatore: scrivi `total += 1`.

### Tradurre un ciclo con contatore: `range`

Per ripetere un’azione un numero stabilito di volte, usiamo `for` insieme a `range`, che descrive una sequenza di interi. Il limite finale è **sempre escluso**:

| Scrittura | Valori attraversati | Significato |
| --- | --- | --- |
| `range(4)` | `0`, `1`, `2`, `3` | Quattro passaggi, a partire da zero |
| `range(2, 5)` | `2`, `3`, `4` | Partenza 2, arresto prima di 5 |
| `range(2, 9, 3)` | `2`, `5`, `8` | Passo di incremento 3 |
| `range(5, 0, -2)` | `5`, `3`, `1` | Passo negativo, sequenza decrescente |

Per esempio, per sommare gli interi da 1 a 4 inclusi:

````python
total = 0
for number in range(1, 5):
    total += number
print(total)  # 10
````

`range` non incrementa una variabile che hai creato fuori dal ciclo: fornisce i valori successivi che `for` assegna a `number`. Riscrivere `number` dentro il corpo non cambia la sequenza dei passaggi successivi.

Il passo deve essere un intero diverso da zero. `range(5, 0)` è vuoto perché il passo predefinito è `1`, mentre per scendere serve un passo negativo. Anche `range(0)` è vuoto. Se vuoi scorrere i valori di una lista, usa direttamente `for value in values`.

### Ripetere finché una condizione è vera: `while`

`while` mantiene la semantica del ciclo con controllo iniziale: verifica la condizione prima di ogni passaggio, esegue il corpo se è vera e termina quando è falsa. L’aggiornamento delle variabili usate nella condizione spetta al programma.

Confronta questo ciclo con il `for` precedente: calcola la stessa somma, ma rende espliciti inizializzazione, controllo e incremento.

````python
total = 0
number = 1
while number <= 4:
    total += number
    number += 1
print(total)
````

La stampa è ancora `10`. Prima dei quattro passaggi `number` vale rispettivamente `1`, `2`, `3`, `4`; dopo l’ultimo incremento vale `5` e la condizione diventa falsa. Se partisse da `5`, il corpo non verrebbe eseguito. Se dimenticassi l’incremento, la condizione resterebbe vera: nel notebook dovresti interrompere l’esecuzione con il pulsante di arresto.

**Verifica di traduzione:** riscrivi con `for` e `range` un ciclo che visita `6`, `4`, `2`, poi si ferma. Quindi riscrivilo con `while`. Il valore zero non deve essere stampato.

<details markdown="1">
<summary>Soluzione della verifica sui cicli</summary>

```python
for number in range(6, 0, -2):
    print(number)

number = 6
while number > 0:
    print(number)
    number -= 2
```

Nel `for`, `0` è il limite escluso e `-2` è il passo. Nel `while`, `number -= 2` equivale a `number = number - 2`; il controllo `> 0` ferma il ciclo quando si arriva a zero. Entrambe le versioni stampano `6`, `4`, `2`.

## 7. Usare e definire funzioni

Le funzioni Python mantengono la distinzione tra definizione e chiamata che già conosci. Abbiamo usato le funzioni predefinite `print` e `len`; ora vediamo come scriverne una nostra, passare gli argomenti e riutilizzare il valore restituito. La scrittura `nome(argomenti)` è una **chiamata**: gli argomenti sono i valori forniti alla funzione, tra parentesi e separati da virgole quando sono più di uno.

`sum(values)` restituisce la somma dei numeri di una lista; `len(values)` restituisce quanti elementi contiene. Possiamo memorizzare questi risultati e usarli per calcolare una media:


````python
values = [10, 20, 30]
total = sum(values)
count = len(values)
print(total)
print(count)
print(total / count)
````

Le stampe danno `60`, `3` e `20.0`. Se vogliamo riutilizzare un nostro calcolo, possiamo definire una funzione con `def`.

Nella definizione seguente:

- `def` introduce la funzione;
- `unit_cost` è il nome scelto per il calcolo del prezzo al metro quadrato;
- `price` e `area` sono i **parametri**, cioè i nomi che useremo nel corpo per riferirci ai dati ricevuti;
- i due punti aprono il blocco e `return` consegna il risultato a chi ha chiamato la funzione.

Come per `if` e `for`, usa Tab o il rientro automatico per scrivere il corpo.


````python
def unit_cost(price, area):
    return price / area
````

Eseguire la definizione rende disponibile `unit_cost`, ma non esegue ancora la divisione e non stampa nulla. Il corpo viene eseguito quando **chiamiamo** la funzione.

Con `unit_cost(200000, 80)`, il primo argomento viene associato al parametro `price` e il secondo ad `area`. La funzione calcola `200000 / 80`, poi `return` termina la chiamata restituendo `2500.0`. Possiamo assegnare quel risultato a una variabile:


````python
first_cost = unit_cost(200000, 80)
second_cost = unit_cost(270000, 120)
print(first_cost)
print(second_cost)
````

Le chiamate restituiscono `2500.0` e `2250.0`. I parametri sono nomi locali alla funzione: ogni chiamata li associa agli oggetti passati come argomenti. Usare `price` come parametro, o assegnargli un altro numero nel corpo, non modifica un’eventuale variabile `price` definita fuori dalla funzione. Questo non significa che Python copi automaticamente gli oggetti ricevuti: vedremo tra poco il caso delle liste.

Possiamo passare gli argomenti **per posizione**, rispettando l’ordine dei parametri, oppure **per nome**, indicando esplicitamente a quale parametro corrispondono:


````python
print(unit_cost(200000, 80))
print(unit_cost(price=200000, area=80))
````

Le due chiamate sono equivalenti. Con i nomi possiamo anche scrivere `unit_cost(area=80, price=200000)`. Qui `price=...` serve a passare un argomento: non è un confronto e non modifica una variabile esterna.

### Restituire un valore non è la stessa cosa che stamparlo

`return` consegna un valore al codice che ha chiamato la funzione e termina quella chiamata. `print` mostra un valore sullo schermo. In `first_cost = unit_cost(200000, 80)`, il risultato viene conservato senza stamparlo; `print(first_cost)` lo rende visibile in un secondo momento.

Una funzione che termina senza `return` restituisce `None`. Anche `print` restituisce `None`: se nella funzione sostituissi `return price / area` con `print(price / area)`, vedresti il numero sullo schermo, ma `first_cost` riceverebbe `None` e non potresti usarlo come numero in altri calcoli.

### Parametri con un valore predefinito

Una funzione può prevedere un valore da usare quando un argomento non viene specificato:

````python
def discounted_price(price, discount=0.10):
    return price * (1 - discount)

print(discounted_price(200000))                 # 180000.0
print(discounted_price(200000, discount=0.20))  # 160000.0
````

Nella **definizione**, `discount=0.10` stabilisce il valore predefinito dello sconto, cioè il 10%. Nella seconda **chiamata**, `discount=0.20` sceglie uno sconto del 20% per quella chiamata. Il parametro obbligatorio `price` viene prima di quello con valore predefinito. Questa distinzione aiuta a leggere le funzioni delle librerie senza dover specificare ogni volta tutti gli argomenti.

### Una funzione può modificare una lista ricevuta

I parametri locali non implicano una copia degli argomenti. Se passi una lista e la funzione chiama `append` su quella lista, la modifica resta visibile all’esterno, come accadeva con i due nomi `prices` e `other_prices`:

````python
def add_price(values, new_price):
    values.append(new_price)

prices = [150000, 200000]
add_price(prices, 180000)
print(prices)  # [150000, 200000, 180000]
````

La funzione modifica la lista ricevuta e non restituisce una nuova lista. Per mantenere intatta quella originale, puoi passare una copia: `add_price(prices[:], 180000)`. In quel caso la modifica riguarda solo la copia temporanea, che questo esempio non conserva. La distinzione da ricordare è tra **riassegnare un nome locale** e **modificare l’oggetto a cui quel nome si riferisce**.

## 8. Dizionari

Un **dizionario** è una raccolta di associazioni tra **chiavi** e **valori**. L’accesso a un valore avviene mediante la chiave corrispondente.

Nell’esempio, le chiavi `"Rooms"`, `"Area"` e `"Price"` identificano le caratteristiche di una casa. Il dizionario è delimitato da parentesi graffe; i due punti separano ciascuna chiave dal valore associato.


````python
house = {"Rooms": 3, "Area": 80.0, "Price": 200000}
print(house["Area"])

house["Rooms"] = 4
print(house["Rooms"])
````

Le stampe producono `80.0` e `4`. `house["Area"]` cerca il valore associato alla chiave `"Area"`; `house["Rooms"] = 4` aggiorna il valore associato a `"Rooms"`. Assegnare a una chiave nuova aggiunge una voce: `house["City"] = "Roma"`.

Una lista si consulta tramite una posizione; un dizionario tramite una chiave. `house[0]` non significa «primo elemento»: cerca la chiave `0`, che qui non esiste. Leggere una chiave assente genera `KeyError`. Anche le maiuscole contano: `"Area"` e `"area"` sono chiavi diverse.

**Prova tu:** aggiungi `"City"` con valore `"Roma"`, poi stampa soltanto la città. Devi ottenere `Roma` usando `print(house["City"])`.

### Riepilogo delle parentesi

| Parentesi | Esempio | Uso |
| --- | --- | --- |
| Tonde `()` | `print(price)` | Chiamata di una funzione |
| Tonde `()` | `(price + 10000) / area` | Raggruppamento di un’espressione |
| Quadre `[]` | `[150000, 200000]` | Creazione di una lista |
| Quadre `[]` | `prices[0]`, `house["Area"]` | Accesso per indice o chiave |
| Graffe `{}` | `{"Rooms": 3}` | Creazione di un dizionario |

I due punti dentro un dizionario separano chiave e valore: non aprono un blocco di istruzioni come dopo `if`. Il significato di un simbolo dipende dal contesto.


## 9. Moduli e importazioni

Python offre molte funzioni già pronte. Alcune, come `print` e `len`, sono disponibili subito; altre sono organizzate in **moduli** che dobbiamo importare. Un modulo raccoglie funzionalità che possiamo usare senza riscriverle.

`math` è un modulo della libreria standard, distribuita con Python: non occorre installarlo. Contiene, per esempio, `sqrt`, che calcola la radice quadrata.

`import math` rende disponibile il nome del modulo. `math.sqrt(25)` accede con il punto alla funzione `sqrt` di quel modulo e la chiama con argomento `25`.


````python
import math

print(math.sqrt(25))
````

Il risultato è `5.0`. La notazione `math.sqrt` identifica la funzione; aggiungere `(25)` la esegue.

### Due forme di importazione che incontrerai nel corso

Possiamo assegnare al modulo un nome locale più breve, detto **alias**:

````python
import math as m
print(m.sqrt(25))  # 5.0
````

`as m` sceglie il nome `m`: il modulo resta lo stesso. È utile riconoscere questa forma perché le librerie del corso usano spesso alias.

Possiamo anche importare una singola funzione:

````python
from math import sqrt
print(sqrt(25))  # 5.0
````

In questo caso `sqrt` è disponibile direttamente e non serve il prefisso `math.`. Le tre forme cambiano i nomi disponibili nel nostro codice, non il calcolo. Prima di usare un nome importato, esegui la cella che contiene l’importazione.

### Attributi: un valore accessibile con il punto

Il punto può dare accesso anche a un valore, chiamato **attributo**. Per esempio, dopo `import math`, `math.pi` contiene un’approssimazione di π:

````python
print(math.pi)       # Circa 3.14159
print(math.sqrt(9))  # 3.0
````

`math.pi` è un numero, quindi non lo chiamiamo con `()`. `sqrt` è una funzione, quindi usiamo le parentesi per eseguirla. Lo stesso criterio di lettura serve per gli oggetti del corso: il punto seleziona un nome; le parentesi indicano una chiamata.

## 10. Leggere e correggere gli errori

Un errore è un’informazione sul punto in cui Python non riesce a proseguire. Parti dall’ultima riga del messaggio: riporta il **tipo di errore** e una descrizione. Poi cerca la riga di codice indicata, correggila e riesegui la cella.

Per esempio, immaginiamo di aver scritto:

```text
price = 180000
print(prcie)
```

Il messaggio termina con una riga simile a:

```text
NameError: name 'prcie' is not defined
```

Python non conosce `prcie`: abbiamo scambiato due lettere. La correzione è `print(price)`. Se il nome è scritto correttamente, controlla di avere eseguito la cella che lo definisce. Dopo un riavvio della sessione occorre ricreare i valori eseguendo le celle in ordine.

| Errore | Esempio o causa | Come intervenire |
| --- | --- | --- |
| `NameError` | Nome non definito | Controlla ortografia e celle eseguite. |
| `SyntaxError` | Parentesi, virgolette o due punti mancanti | Controlla la riga indicata e quella precedente. |
| `IndentationError` | Blocco senza rientro o rientro inatteso | Riallinea le righe usando l’editor. |
| `TabError` | Tabulazioni e spazi mescolati in modo incoerente | Uniforma l’indentazione con l’editor, poi usa Tab per i rientri. |
| `IndexError` | `prices[3]` su una lista di tre elementi | Gli indici validi partono da zero e arrivano a `len(prices) - 1`. |
| `KeyError` | `house["area"]` quando la chiave è `"Area"` | Controlla il nome esatto della chiave. |
| `TypeError` | `"80" + 20` | Decidi se vuoi sommare numeri o unire testi e converti il valore necessario. |
| `ValueError` | `int("ciao")` | Il testo deve rappresentare un numero convertibile. |
| `ZeroDivisionError` | Divisione per zero | Controlla il denominatore, per esempio una superficie o la lunghezza di una lista vuota. |

Se un errore avviene durante l’esecuzione, le istruzioni successive della stessa cella non vengono eseguite; quelle precedenti possono aver già modificato i valori in memoria. Dopo la correzione riesegui la cella completa, controllando le variabili da cui parte.
