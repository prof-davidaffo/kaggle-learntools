## 4. Indentazione e blocchi

Un **blocco** è un gruppo di istruzioni che appartiene a una condizione, a un ciclo o a una funzione. In linguaggi come C e Java il blocco viene racchiuso tra parentesi graffe, `{ ... }`. **In Python sono gli spazi all'inizio delle righe a indicare quali istruzioni appartengono al blocco.** Questi spazi si chiamano *indentazione* o *rientro* e cambiano il significato del programma.

La regola da usare negli esercizi è:

1. Scrivi i **due punti `:`** alla fine di una riga che apre un blocco, per esempio `if`, `else`, `for` o `def`.
2. Vai a capo e rientra di **quattro spazi** per scrivere le istruzioni del blocco.
3. Mantieni lo **stesso rientro** per tutte le istruzioni dello stesso blocco.
4. Per terminare il blocco, torna al rientro precedente. Non si scrive una parentesi graffa di chiusura.

Quattro spazi sono la convenzione Python che seguiremo nel corso: un rientro coerente di altra ampiezza può essere valido, ma non mescolare stili. Usa spazi anziché tabulazioni; Jupyter normalmente inserisce il rientro dopo i due punti.

### Dentro e fuori dal blocco


````python
price = 180000

if price < 200000:
    print("Price below 200000")
    print("This house is within budget")

print("Price check finished")
````

Le due righe rientrate appartengono all'`if`: vengono eseguite soltanto se `price < 200000` è vero. L'ultimo `print()` è tornato al margine sinistro, quindi è fuori dal blocco e viene eseguito comunque.

**Prova:** imposta `price = 250000` e riesegui la cella. Comparirà soltanto `Price check finished`. Poi sposta l'ultimo `print()` dentro il blocco aggiungendo quattro spazi: con quel prezzo non verrà stampato nulla. L'indentazione ha cambiato il comportamento del programma.

### Blocchi annidati

Un blocco può contenerne un altro. Ogni livello aggiunge quattro spazi: qui il corpo del `for` ne ha quattro, mentre il corpo dell'`if` interno ne ha otto.


````python
prices = [150000, 250000, 180000]

for price in prices:
    if price < 200000:
        print("Within budget:", price)
    print("Checked:", price)

print("All houses checked")
````

- `Within budget` viene stampato soltanto per i prezzi sotto 200000, perché è nel blocco dell'`if`.
- `Checked` viene stampato per ogni prezzo: è dentro il `for`, ma fuori dall'`if`.
- `All houses checked` viene stampato una sola volta alla fine: è fuori da entrambi i blocchi.

Lo stesso vale per `else`: deve essere allineato con l'`if` a cui si riferisce; le sue istruzioni vanno rientrate di quattro spazi.

### Gli errori di rientro

Questo esempio è **errato**: dopo i due punti manca il rientro.

```text
if price < 200000:
print("Within budget")
```

Python segnala `IndentationError: expected an indented block`. Per correggerlo, aggiungi quattro spazi prima di `print()`.

Anche rientrare senza aver aperto un blocco può produrre `IndentationError: unexpected indent`. Mescolare spazi e tabulazioni può causare `TabError`. Se una riga rientra a un livello diverso da quelli già aperti, puoi ricevere `IndentationError: unindent does not match any outer indentation level`.

Un rientro sbagliato può però essere **sintatticamente valido**: nell'esempio precedente spostare `Checked` dentro l'`if` non dà errore, ma fa saltare il messaggio per i prezzi fuori budget. Controlla quindi anche il risultato, non soltanto l'assenza di errori.

### Quali parentesi si usano in Python?

Python usa le parentesi, ma **non usa le graffe per delimitare i blocchi**.

| Simbolo | Uso | Esempio |
| --- | --- | --- |
| `()` | Chiamare una funzione o raggruppare un calcolo | `print(price)`, `(price + 10000) / area` |
| `[]` | Creare una lista o selezionare elementi e colonne | `[2, 3, 4]`, `prices[0]`, `data["Price"]` |
| `{}` | Creare un dizionario; si usano anche per gli insiemi | `{"Rooms": 3, "Area": 80}` |

Puoi andare a capo dentro parentesi aperte per rendere un'espressione più leggibile. Quel rientro serve a impaginare l'espressione: non apre un blocco `if`, `for` o `def`.


````python
feature_names = [
    "Rooms",
    "Area",
    "YearBuilt",
]
print(feature_names)
````

## 5. Condizioni, cicli e funzioni


`if` sceglie quali istruzioni eseguire in base a una condizione. `else` gestisce il caso alternativo. `for` ripete il blocco per ogni elemento di una sequenza.


````python
error = 12000

if error < 15000:
    print("Error below threshold")
else:
    print("Error above threshold")

for depth in [2, 5, 10]:
    print("Depth:", depth)
````

Una **funzione** raccoglie operazioni riutilizzabili. Riceve argomenti e restituisce un risultato con `return`. `print()` mostra un valore, ma non lo restituisce al chiamante.


````python
def absolute_error(actual, predicted):
    return abs(actual - predicted)

error = absolute_error(actual=200000, predicted=185000)
print(error)
````

`actual=...` e `predicted=...` sono argomenti passati per nome. Troverai la stessa sintassi in `train_test_split(..., random_state=42)` e nei costruttori dei modelli.

## 6. Import, oggetti e metodi

Con il punto si accede a un attributo o a un metodo di un oggetto:

## 7. Le operazioni pandas necessarie

Un **DataFrame** è una tabella con righe e colonne. Una **Series** rappresenta, tra le altre cose, una singola colonna. Per provare la sintassi usiamo questi quattro record di esempio, senza scaricare file.


````python
data = pd.DataFrame({
    "Rooms": [2, 3, 4, 3],
    "Area": [60.0, 80.0, 120.0, None],
    "Price": [150000, 200000, 270000, 190000],
})
data.head()
````

`shape` restituisce una coppia `(numero di righe, numero di colonne)`. `columns` elenca i nomi delle colonne; `describe()` riassume quelle numeriche.


````python
print(data.shape)
print(data.columns)
data.describe()
````

Per selezionare una colonna si usa il suo nome. Per selezionarne più di una si passa una lista: da qui le doppie parentesi quadre.


````python
y = data["Price"]
X = data[["Rooms", "Area"]]

print(type(y))
print(type(X))
print(X.shape)
````

Nel corso **`X` indica le feature**, cioè le colonne usate per prevedere; **`y` indica il target**, cioè ciò che vuoi prevedere. Il prezzo non va incluso in `X` se è il valore da prevedere.

Un confronto su una colonna restituisce un valore booleano per ogni riga. Puoi usarlo per filtrare la tabella. `.iloc` seleziona per posizione; `.loc` può combinare un filtro sulle righe e i nomi delle colonne.


````python
large_houses = data.loc[data["Area"] >= 80, ["Rooms", "Price"]]
print(large_houses)
print(data.iloc[0])
````

Per contare i dati mancanti usa `isna().sum()`. Negli esempi iniziali puoi eliminare le righe che hanno dati mancanti nelle colonne necessarie. È una scelta semplificata: nei dataset reali può escludere molti esempi utili.


````python
print(data.isna().sum())
clean_data = data.dropna(subset=["Rooms", "Area", "Price"])
X = clean_data[["Rooms", "Area"]]
y = clean_data["Price"]
print(X.shape, y.shape)
````

## 8. Array NumPy e grafici

Un **array NumPy** contiene dati su cui puoi fare operazioni numeriche elemento per elemento. Su una lista Python, invece, `* 2` ripete gli elementi.


````python
values = [10, 20, 30]
array = np.array(values)
print(values * 2)
print(array * 2)

matrix = np.array([[2, 60], [3, 80], [4, 120]])
print(matrix.shape)
print(matrix[:, 0])
````

`matrix[:, 0]` seleziona tutte le righe della prima colonna. Nelle reti neurali la forma degli input conta: una tabella con due feature ha forma `(numero di esempi, 2)`. `X.shape[1]` restituisce il numero di feature.

## 9. Leggere gli errori

Quando una cella fallisce, leggi prima l'ultima riga del messaggio e poi il punto indicato nel codice.

| Errore | Cosa controllare |
| --- | --- |
| `NameError` | Hai definito la variabile ed eseguito la cella? Il nome è scritto correttamente? |
| `SyntaxError` o `IndentationError` | Mancano parentesi, virgolette, due punti o rientri? |
| `KeyError` | Il nome della colonna esiste in `data.columns`? |
| `FileNotFoundError` | Il file esiste nel percorso indicato? Hai eseguito la preparazione? |
| `ModuleNotFoundError` | La libreria è installata nell'ambiente del kernel selezionato? |
| `ValueError` | Dimensioni e valori degli input sono compatibili con l'operazione? |
