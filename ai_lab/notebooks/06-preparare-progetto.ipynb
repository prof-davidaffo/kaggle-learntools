## Obiettivo
Definire target, feature, unità e divisione dei dati. Il dataset di riferimento rappresenta due specie **fittizie** di fiori; serve a studiare il processo, non è utilizzabile per botanica reale. Nessuna dipendenza esterna.


```python
import csv
import json
import math
import random
from pathlib import Path
from statistics import mean, pstdev

rng = random.Random(27)
rows = []
for index in range(600):
    label = index % 2
    length = min(9.5, max(0.5, rng.gauss(3.0 + 3.5 * label, 1.0)))
    width = min(5.0, max(0.2, rng.gauss(1.0 + 1.5 * label, 0.65)))
    rows.append(([length, width], label))
rng.shuffle(rows)
train_rows, valid_rows, test_rows = rows[:360], rows[360:480], rows[480:]
feature_mean = [mean(row[0][j] for row in train_rows) for j in range(2)]
feature_scale = [pstdev(row[0][j] for row in train_rows) for j in range(2)]


def transform(features):
    return [(value - center) / scale for value, center, scale in
            zip(features, feature_mean, feature_scale)]


def sigmoid(value):
    return 1 / (1 + math.exp(-max(-40, min(40, value))))

print('Synthetic examples:', len(rows), 'Features:', ['length', 'width'])
```

```python
for name, split in [("train",train_rows),("validation",valid_rows),("test",test_rows)]:
    print(name, "Count:",len(split),"Positive fraction:",mean(label for _,label in split))
assert len({id(row) for row in train_rows} & {id(row) for row in test_rows}) == 0
print("Scaler fitted on training only:",feature_mean,feature_scale)
```
