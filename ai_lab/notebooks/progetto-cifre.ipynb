## Obiettivi e dati
Usa `load_digits` di scikit-learn: immagini 8×8 di cifre da 0 a 9, distribuite con la libreria, senza download aggiuntivo dopo l’installazione. Fonte: UCI Optical Recognition of Handwritten Digits, E. Alpaydin e C. Kaynak. Leggi `digits.DESCR` e la documentazione scikit-learn per provenienza e condizioni prima di ridistribuire. CPU sufficiente. Esecuzione locale non verificata.


```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix

digits = load_digits()
X_train, X_test, y_train, y_test = train_test_split(digits.data, digits.target, test_size=0.2, stratify=digits.target, random_state=42)
X_train, X_valid, y_train, y_valid = train_test_split(X_train, y_train, test_size=0.25, stratify=y_train, random_state=42)
model = make_pipeline(StandardScaler(), LogisticRegression(max_iter=2000, random_state=42))
model.fit(X_train, y_train)
print("Validation accuracy:",accuracy_score(y_valid,model.predict(X_valid)))
print("Final test accuracy:",accuracy_score(y_test,model.predict(X_test)))
print(confusion_matrix(y_test,model.predict(X_test)))
print("First prediction:",model.predict(X_test[:1]))
```

## Esercizi
Prima del test, confronta un classificatore semplice con un MLP sul validation set. Esamina almeno cinque errori e gli effetti della normalizzazione.

## Interfaccia
Una griglia 8×8 o un’immagine ridimensionata e normalizzata come il training. Un disegno esterno può avere una distribuzione diversa: non promettere la stessa accuracy. Gradio può essere eseguito separatamente dal sito.
